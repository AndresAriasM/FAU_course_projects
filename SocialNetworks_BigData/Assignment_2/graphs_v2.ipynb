```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
import random
import ast

# Same subsample as Project Part #1, BFS from a random seed node, first 5000 nodes
path_edges = '/Users/andresarias/Documents/FAU/SocialNetworks_BigData/Assignment_1/data/edges.csv'
path_nodes = '/Users/andresarias/Documents/FAU/SocialNetworks_BigData/Assignment_1/data/nodes.csv'

edges = pd.read_csv(path_edges)
nodes = pd.read_csv(path_nodes)

G = nx.from_pandas_edgelist(edges, source='id_0', target='id_1')

random.seed(42)
start_node = random.choice(list(G.nodes))
bfs_tree = nx.bfs_tree(G, source=start_node)
sample_nodes = list(bfs_tree.nodes)[:5000]
G_sample = G.subgraph(sample_nodes).copy()

# Drop duplicates in the nodes dataset
nodes_unique = nodes.drop_duplicates(subset='spotify_id', keep='first')
id_to_name = dict(zip(nodes_unique['spotify_id'], nodes_unique['name']))

print("Start node:", start_node)
print(G_sample)
```

## 1. Compute degree centrality for all nodes

```python
degree_centrality = nx.degree_centrality(G_sample)

print("Degree centrality computed for", len(degree_centrality), "nodes")
print("Some of them", list(degree_centrality.items())[:2])
```

```python
degree_centrality
```

## 2. Print out the top 10 ARTIST NAMES (not node ids) with the highest degree centrality

```python
# Degree centrality is a dictionary, so i will use .items() to obtain key:value with an iteration
# sort to order all the data collection and key=lambda x: x[1] to order by values and not with keys (IDs)
top10_degree = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)[:10]

print("Top 10 artists by degree centrality")
for i, (node_id, score) in enumerate(top10_degree, start=1): # enumerate to avoid creating a counter
    print((i, id_to_name[node_id], score))
```
